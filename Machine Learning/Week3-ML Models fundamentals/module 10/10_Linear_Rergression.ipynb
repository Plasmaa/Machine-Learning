```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
hours = np.array([2,4,6,8,10]).reshape(-1,1)
marks = np.array([55,60,70,80,88])

hours,marks
```

```python
# Now training the brain / model - using scikit learn
model = LinearRegression()
model.fit(hours, marks)
```

```python
print("m: ", model.coef_[0])
print("c: ", model.intercept_)

# marks = 4.3xhours + 44.8
```

```python
# Visualization
plt.scatter(hours, marks, color='blue', label="Actual Data")

predicted_marks = model.predict(hours)
plt.plot(hours, predicted_marks, color='red', label="Predicted Line")

plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks for linear regression")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Prediction

new_hour = np.array([2]).reshape(-1, 1)
prediction = model.predict(new_hour)
print("Predicted Mark: ", prediction[0])
```
